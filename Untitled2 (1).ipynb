```python
import pandas as pd
import time
import math



##Research question: Do higher scoring players also have high efficiency


##cleaning
df = pd.read_csv("all_seasons.csv")
if "Unnamed: 0" in df.columns:
    df = df.drop(columns=["Unnamed: 0"])
df["college"] = df["college"].fillna("No College Listed")

df = df[df["gp"] >= 50]
df = df[df["pts"] >= 0]
df = df[df["ts_pct"] >= 0]
df = df[df["ts_pct"] <= 1]

##print(df.head())
##print(df.info())


##structures

p_array = []

for i in range(len(df)):
    row = df.iloc[i]

    player = {
        "player_name": row["player_name"],
        "team": row["team_abbreviation"],
        "season": row["season"],
        "gp": row["gp"],
        "pts": row["pts"],
        "ts_pct": row["ts_pct"]
    }

    p_array.append(player)

p_hash = {}

for player in p_array:
    name = player["player_name"]

    if name not in p_hash:
        p_hash[name] = []

    p_hash[name].append(player)


low_total_ts = 0
low_count = 0

medium_total_ts = 0
medium_count = 0

high_total_ts = 0
high_count = 0

for player in p_array:
    if player["pts"] < 10:
        low_total_ts += player["ts_pct"]
        low_count += 1

    elif player["pts"] < 20:
        medium_total_ts += player["ts_pct"]
        medium_count += 1

    else:
        high_total_ts += player["ts_pct"]
        high_count += 1


low_avg_ts = low_total_ts / low_count
medium_avg_ts = medium_total_ts / medium_count
high_avg_ts = high_total_ts / high_count

print("Array Results")

print("Low scorers(<10) average TS%:", low_avg_ts)
print("Medium scorers(10<x<20) average TS%:", medium_avg_ts)
print("High scorers(>20) average TS%:", high_avg_ts)

print("Low scorer count:", low_count)
print("Medium scorer count:", medium_count)
print("High scorer count:", high_count)
low_total_ts_hash = 0
low_count_hash = 0

medium_total_ts_hash = 0
medium_count_hash = 0

high_total_ts_hash = 0
high_count_hash = 0

for name in p_hash:
    seasons = p_hash[name]

    for player in seasons:
        if player["pts"] < 10:
            low_total_ts_hash += player["ts_pct"]
            low_count_hash += 1

        elif player["pts"] < 20:
            medium_total_ts_hash += player["ts_pct"]
            medium_count_hash += 1

        else:
            high_total_ts_hash += player["ts_pct"]
            high_count_hash += 1


low_avg_ts_hash = low_total_ts_hash / low_count_hash
medium_avg_ts_hash = medium_total_ts_hash / medium_count_hash
high_avg_ts_hash = high_total_ts_hash / high_count_hash

print()
print("Hash Table Results:")
print("Low scorers(<10) average TS%:", low_avg_ts_hash)
print("Medium scorers(10<x<20) average TS%:", medium_avg_ts_hash)
print("High scorers average(>20) TS%:", high_avg_ts_hash)

print("Low scorer count:", low_count_hash)
print("Medium scorer count:", medium_count_hash)
print("High scorer count:", high_count_hash)
start = time.perf_counter()

low_total_ts = 0
low_count = 0

medium_total_ts = 0
medium_count = 0

high_total_ts = 0
high_count = 0

for player in p_array:
    if player["pts"] < 10:
        low_total_ts += player["ts_pct"]
        low_count += 1

    elif player["pts"] < 20:
        medium_total_ts += player["ts_pct"]
        medium_count += 1

    else:
        high_total_ts += player["ts_pct"]
        high_count += 1

low_avg_ts = low_total_ts / low_count
medium_avg_ts = medium_total_ts / medium_count
high_avg_ts = high_total_ts / high_count

end = time.perf_counter()

array_analysis_time = end - start

print("Array analysis time:", array_analysis_time)
start = time.perf_counter()

start = time.perf_counter()

low_total_ts_hash = 0
low_count_hash = 0

medium_total_ts_hash = 0
medium_count_hash = 0

high_total_ts_hash = 0
high_count_hash = 0

for name in p_hash:
    seasons = p_hash[name]

    for player in seasons:
        if player["pts"] < 10:
            low_total_ts_hash += player["ts_pct"]
            low_count_hash += 1

        elif player["pts"] < 20:
            medium_total_ts_hash += player["ts_pct"]
            medium_count_hash += 1

        else:
            high_total_ts_hash += player["ts_pct"]
            high_count_hash += 1

low_avg_ts_hash = low_total_ts_hash / low_count_hash
medium_avg_ts_hash = medium_total_ts_hash / medium_count_hash
high_avg_ts_hash = high_total_ts_hash / high_count_hash

end = time.perf_counter()

hash_analysis_time = end - start

print("Hash table analysis time:", hash_analysis_time)





```
